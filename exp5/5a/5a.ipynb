```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("sales.csv")

# Display dataset
print(df)
```

```python

# Store data
months = df["Month"]
sales = df["Sales"]
```

```python
# Program 1: Line Chart
plt.plot(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
# Program 2: Bar Chart
plt.bar(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
# Program 3: Pie Chart
plt.pie(sales, labels=months, autopct="%1.1f%%")
plt.title("Sales Distribution")
plt.show()
```

```python
# Program 4: Histogram
plt.hist(sales, bins=5)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()
```

```python
# Program 5: Scatter Plot
plt.scatter(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python









# Program 7: Multiple Line Charts
profit = [20, 30, 40, 35, 50, 60]

plt.plot(months, sales, label="Sales")
plt.plot(months, profit, label="Profit")
plt.title("Sales and Profit")
plt.xlabel("Months")
plt.ylabel("Value")
plt.legend()
plt.show()
```
