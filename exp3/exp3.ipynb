```python
import pandas as pd
df = pd.read_csv('fs.csv')
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
# Calculate mean of numerical columns
mean_values = df.mean(numeric_only=True)

print(mean_values)
```

```python
# Replace null values with column mean
df.fillna(mean_values, inplace=True)
```

```python
df["Revenue"] = df["Revenue"].fillna(df["Revenue"].mean())
df["Gross Profit"] = df["Gross Profit"].fillna(df["Gross Profit"].mean())
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated())
```

```python
Q1 = df['Revenue'].quantile(0.25)
Q3 = df['Revenue'].quantile(0.75)

IQR = Q3 - Q1

outliers = df[(df['Revenue'] < Q1 - 1.5*IQR) | (df['Revenue'] > Q3 + 1.5*IQR)]

print(outliers)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.boxplot(df['Revenue'].dropna())
plt.title("Revenue Outlier Detection")
plt.ylabel("Revenue")
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['Revenue']] = scaler.fit_transform(df[['Revenue']])

print(df[['Revenue']])
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df['Category'] = encoder.fit_transform(df['Category'])

print(df[['Category']])
```
