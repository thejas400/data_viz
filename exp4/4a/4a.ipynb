```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read dataset
df = pd.read_csv("stud_data.csv")

# Display dataset
print(df)
```

```python
# Display first five records
print(df.head())

# Display shape
print(df.shape)

# Display data types
print(df.dtypes)
```

```python
# Display statistical summary
print(df.describe())

# Filter students with marks greater than 80
print(df[df["Marks"] > 80])

# Filter students with attendance greater than 90
print(df[df["Attendance"] > 90])
```

```python
# Sort students by marks
print(df.sort_values("Marks", ascending=False))

# Sort students by attendance
print(df.sort_values("Attendance", ascending=False))

# Department-wise average marks
print(df.groupby("Department")["Marks"].mean())

# Department-wise average attendance
print(df.groupby("Department")["Attendance"].mean())
```

```python

# Calculate mean
print("Mean Marks:", df["Marks"].mean())

# Calculate median
print("Median Marks:", df["Marks"].median())

# Calculate minimum
print("Minimum Marks:", df["Marks"].min())

# Calculate maximum
print("Maximum Marks:", df["Marks"].max())

# Calculate standard deviation
print("Standard Deviation:", df["Marks"].std())

# Display correlation matrix
print(df[["Marks", "Attendance"]].corr())

# Display correlation between marks and attendance
print("Correlation:", df["Marks"].corr(df["Attendance"]))

```

```python
# Bar chart
plt.bar(df["Name"], df["Marks"])
plt.title("Student Marks")
plt.xlabel("Name")
plt.ylabel("Marks")
plt.show()
```

```python

# Line chart
plt.plot(df["Name"], df["Attendance"], marker="o")
plt.title("Student Attendance")
plt.xlabel("Name")
plt.ylabel("Attendance")
plt.show()
```

```python

# Scatter plot
plt.scatter(df["Attendance"], df["Marks"])
plt.title("Marks vs Attendance")
plt.xlabel("Attendance")
plt.ylabel("Marks")
plt.show()
```

```python
# Histogram
plt.hist(df["Marks"], bins=5)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()
```

```python


# Box plot
plt.boxplot(df["Marks"])
plt.title("Marks Box Plot")
plt.ylabel("Marks")
plt.show()
```
