```python
import pandas as pd

df = pd.read_csv("fs.csv")
df.columns = df.columns.str.strip()

# Find the exact column containing 'Net Inc'
net_inc_col = [c for c in df.columns if "Net" in c][0]

print(f"--- Net Income Statistics (Using: '{net_inc_col}') ---")
print("Mean =", df[net_inc_col].mean())
print("Median =", df[net_inc_col].median())
print("Mode =\n", df[net_inc_col].mode().to_string())
print("Standard Deviation =", df[net_inc_col].std())
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("fs.csv")
df.columns = df.columns.str.strip()

# Dynamic column selection
market_col = [c for c in df.columns if "Market" in c][0]
year_col = [c for c in df.columns if "Year" in c][0]

# Ensure chronological ordering
df_sorted = df.sort_values(by=year_col)

plt.figure(figsize=(8, 4))
plt.plot(
    df_sorted[year_col],
    df_sorted[market_col],
    marker="o",
    color="blue",
    linewidth=2,
)
plt.title("Market Cap Evolution (2017 - 2022)")
plt.xlabel("Year")
plt.ylabel("Market Capitalization")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv("fs.csv")
df.columns = df.columns.str.strip()

# Dynamic column selection
year_col = [c for c in df.columns if "Year" in c][0]
rev_col = [c for c in df.columns if "Revenue" in c][0]
net_inc_col = [c for c in df.columns if "Net" in c][0]

df_sorted = df.sort_values(by=year_col)

# Set positions for the bars
x = np.arange(len(df_sorted[year_col]))
width = 0.35

plt.figure(figsize=(9, 5))
plt.bar(x - width / 2, df_sorted[rev_col], width, label="Revenue", color="teal")
plt.bar(
    x + width / 2,
    df_sorted[net_inc_col],
    width,
    label="Net Income",
    color="orange",
)

plt.title("Revenue vs Net Income Comparison")
plt.xlabel("Year")
plt.ylabel("Financial Amount")
plt.xticks(x, df_sorted[year_col])
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("fs.csv")
df.columns = df.columns.str.strip()

# Dynamic column selection
net_inc_col = [c for c in df.columns if "Net" in c][0]
inf_col = [c for c in df.columns if "Inflation" in c][0]

plt.figure(figsize=(7, 4.5))
plt.scatter(df[inf_col], df[net_inc_col], color="red", s=100, edgecolors="black")

plt.title("Impact of Inflation on Net Income")
plt.xlabel("Inflation Rate (%)")
plt.ylabel("Net Income")
plt.grid(True, linestyle=":", alpha=0.6)
plt.show()
```
